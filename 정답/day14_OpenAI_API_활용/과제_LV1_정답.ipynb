## 1. 첫 대화 — 모델에게 물어보기
**배경**: 가장 기본은 모델에게 질문을 보내 답을 받는 것입니다.

**요구사항**:
- `client.chat.completions.create` 로 `model='gpt-4o-mini'` 에게 **"파이썬은 어떤 언어인가요? 한 문장으로."** 를 `user` 메시지로 보내세요.
- 답변 문자열(`resp.choices[0].message.content`)을 변수 **`answer`** 에 담으세요.

<details><summary>힌트</summary>

```text
접근방법:
- messages 는 [{'role':'user','content': 질문}] 형태의 리스트다.

세부구현:
1. create 를 호출해 결과를 resp 에 담는다(model, messages 지정).
2. resp 의 choices 첫 번째의 message 의 content 를 answer 에 담는다.
3. answer 를 출력한다.
```

</details>

````python
resp = client.chat.completions.create(
    model='gpt-4o-mini',
    # 딕셔너리 하나가 대화 한 줄이다 — role 로 누가 말했는지, content 로 무슨 말인지 적는다.
    messages=[{'role': 'user', 'content': '파이썬은 어떤 언어인가요? 한 문장으로.'}])
# choices 가 리스트인 것은 답을 여러 개 받을 수도 있기 때문이다. 평소에는 [0] 하나만 쓴다.
answer = resp.choices[0].message.content
print(answer)
````

````python
# [자가채점]
assert isinstance(answer, str) and len(answer.strip()) > 0
print('✅ 통과!')
````

### 해설

`messages` 는 딕셔너리들의 리스트입니다. 답은 항상 `choices[0].message.content` 에 문자열로 옵니다. 흔한 실수: `messages` 에 리스트가 아닌 문자열을 바로 넣는 것.

## 2. Responses API 로 물어보기
**배경**: 같은 질문을 더 간단한 **Responses API** 로도 할 수 있습니다.

**요구사항**:
- `client.responses.create` 로 `model='gpt-4o-mini'`, `input='자바스크립트는 어떤 언어인가요? 한 문장으로.'` 를 호출하세요.
- 답변 문자열(`resp.output_text`)을 변수 **`answer_r`** 에 담으세요.

<details><summary>힌트</summary>

```text
접근방법:
- Responses 는 messages 대신 input 에 문자열을 바로 준다.

세부구현:
1. responses.create(model, input) 을 호출해 resp 에 담는다.
2. resp 의 output_text 를 answer_r 에 담아 출력한다.
```

</details>

````python
resp = client.responses.create(
    model='gpt-4o-mini',
    # 대화 이력을 쌓을 필요가 없는 한 번짜리 질문이면 이쪽이 짧다(리스트를 만들지 않아도 된다).
    input='자바스크립트는 어떤 언어인가요? 한 문장으로.')
answer_r = resp.output_text
print(answer_r)
````

### 해설

## 3. 역할 지정 — 번역기로 만들기
**배경**: `system` 메시지로 모델에게 역할을 주면 그 역할대로 답합니다.

**요구사항**:
- `system` 에 **'입력한 한국어를 영어로 번역만 해. 설명은 하지 마.'**, `user` 에 **'저는 매일 아침 커피를 마셔요.'** 를 넣어 `gpt-4o-mini` 로 호출하세요.
- 답변 문자열을 변수 **`answer_sys`** 에 담으세요.

<details><summary>힌트</summary>

```text
접근방법:
- messages 리스트에 system 과 user 두 개를 순서대로 넣는다.

세부구현:
1. messages=[{'role':'system',...}, {'role':'user',...}] 로 만든다.
2. create 호출 후 content 를 answer_sys 에 담아 출력한다.
```

</details>

````python
resp = client.chat.completions.create(
    model='gpt-4o-mini',
    messages=[
        # system 은 대화 전체에 거는 규칙이다 — user 에 그때그때 적는 부탁과 달리 뒤 대화에서도 유지된다.
        {'role': 'system', 'content': '입력한 한국어를 영어로 번역만 해. 설명은 하지 마.'},
        # 그래서 user 에는 지시를 빼고 번역할 문장만 남길 수 있다.
        {'role': 'user', 'content': '저는 매일 아침 커피를 마셔요.'}])
answer_sys = resp.choices[0].message.content
print(answer_sys)
````

````python
# [자가채점]
import re
assert isinstance(answer_sys, str) and len(answer_sys.strip()) > 0
# 번역기 역할이 실제로 먹혔는지 — 영어 문장이 돌아와야 한다
assert re.search(r'[A-Za-z]{3,}', answer_sys), 'system 역할이 안 먹혔습니다 - 영어 번역 결과여야 합니다'
print('✅ 통과!')
````

### 해설

`system` 은 보통 맨 앞에 한 번 넣어 역할·규칙을 정합니다. 번역·요약 같은 작업에선 "설명하지 마" 같은 제약을 주면 군더더기가 줄어듭니다.

## 4. 역할 지정 — 분류기로 만들기
**배경**: 3번(번역)과 같은 'system 역할 주기'지만, 이번엔 **판별(분류)** 을 시킵니다 — 글을 바꾸는 게(변환) 아니라 **정해진 라벨 하나를 고르게** 합니다(같은 개념·다른 유형).

**요구사항**:
- `system` 에 **'너는 리뷰 감정 분류기야. 리뷰를 긍정/부정/중립 한 단어로만 답해.'**, `user` 에 아래 `review` 를 넣어 `gpt-4o-mini` 로 호출하세요.
- 답변 문자열을 변수 **`answer_label`** 에 담으세요.

```python
review = '기대했는데 배송도 느리고 품질도 영 별로였어요.'
```

<details><summary>힌트</summary>

```text
접근방법:
- 3번처럼 system+user 로 호출하되, system 에 '분류'를 지시하고 라벨 후보를 정해 준다.

세부구현:
1. 먼저 review 변수를 만든다.
2. system=분류 지시(라벨 후보 포함), user=review 로 create 호출한다.
3. content 를 answer_label 에 담아 출력한다.
```

</details>

````python
review = '기대했는데 배송도 느리고 품질도 영 별로였어요.'
resp = client.chat.completions.create(
    model='gpt-4o-mini',
    messages=[
        # 분류를 시킬 때는 라벨 후보와 '한 단어로만'을 함께 못박는다.
        #  형식을 안 정하면 "부정적인 리뷰네요" 같은 문장이 와서 나중에 세지 못한다.
        {'role': 'system', 'content': '너는 리뷰 감정 분류기야. 리뷰를 긍정/부정/중립 한 단어로만 답해.'},
        {'role': 'user', 'content': review}])
answer_label = resp.choices[0].message.content
print(answer_label)
````

````python
# [자가채점]
assert isinstance(answer_label, str) and len(answer_label.strip()) > 0
# 분류기 역할이 실제로 먹혔는지 — 지정한 라벨 중 하나가 나와야 한다
assert any(lab in answer_label for lab in ('긍정', '부정', '중립')), \
    'system 에 준 라벨(긍정/부정/중립) 중 하나가 답에 들어 있어야 합니다'
print('✅ 통과!')
````

### 해설

3·4번은 'system 으로 역할 주기'라는 같은 개념을 **다른 유형**으로 연습합니다 — 3번은 입력을 바꾸는 **변환(번역)**, 4번은 정해진 라벨을 고르는 **판별(분류)**. system 에 라벨 후보를 명시하면 답이 그 안에서 나옵니다(프롬프트로 출력을 좁히는 감각).

## 5. 파라미터 — temperature 로 다양성 조절
**배경**: 창작에는 `temperature` 를 높여 다양한 답을 얻습니다.

**요구사항**:
- 온도 값을 변수 **`temp = 1.2`** 에 담고, `user='재미있는 카페 이름을 하나만 지어줘. 이름만 답해.'` 를 `gpt-4o-mini`, **`temperature=temp`** 로 호출하세요.
- 답변 문자열을 변수 **`answer_t`** 에 담으세요.

<details><summary>힌트</summary>

```text
접근방법:
- 온도 값을 변수에 담아 create 의 temperature 인자로 넘긴다.

세부구현:
1. temp 에 1.2 를 담는다.
2. create(model, messages, temperature=temp) 로 호출한다.
3. content 를 answer_t 에 담아 출력한다.
```

</details>

````python
temp = 1.2   # 1 보다 크면 평소 잘 안 고르던 단어까지 후보에 들어온다
resp = client.chat.completions.create(
    model='gpt-4o-mini',
    messages=[{'role': 'user', 'content': '재미있는 카페 이름을 하나만 지어줘. 이름만 답해.'}],
    # 이름 짓기처럼 '매번 달라야 좋은' 일에만 올린다. 여러 번 실행해 보면 답이 바뀐다.
    temperature=temp)
answer_t = resp.choices[0].message.content
print(answer_t)
````

````python
# [자가채점]
assert temp == 1.2
assert isinstance(answer_t, str) and len(answer_t.strip()) > 0
print('✅ 통과!')
````

### 해설

`temperature` 를 높이면 덜 유력한 토큰도 자주 뽑혀 답이 다양해집니다. 사실·번역이라면 반대로 `temperature=0` 이 좋습니다.

## 6. 파라미터 — max_tokens 로 길이 제한
**배경**: 답이 길어 요금·속도가 부담이면 `max_tokens` 로 최대 길이를 제한합니다(5번과 같은 '파라미터' 개념).

**요구사항**:
- 최대 길이를 변수 **`limit = 25`** 에 담고, `user='머신러닝이 무엇인지 자세히 설명해줘.'` 를 `gpt-4o-mini`, **`max_tokens=limit`** 로 호출하세요.
- 답변 문자열을 변수 **`answer_m`** 에 담으세요.

<details><summary>힌트</summary>

```text
접근방법:
- 최대 토큰 값을 변수에 담아 create 의 max_tokens 인자로 넘긴다.

세부구현:
1. limit 에 25 를 담는다.
2. create(model, messages, max_tokens=limit) 로 호출한다.
3. content 를 answer_m 에 담아 출력한다(짧게 끊긴다).
```

</details>

````python
limit = 25
resp = client.chat.completions.create(
    model='gpt-4o-mini',
    messages=[{'role': 'user', 'content': '머신러닝이 무엇인지 자세히 설명해줘.'}],
    # max_tokens 는 '짧게 써라'가 아니라 '여기서 끊어라'다 —
    #  그래서 문장 중간에서 잘린 답이 나오는 것이 정상이다(짧게 받고 싶으면 지시문에 적어야 한다).
    max_tokens=limit)
answer_m = resp.choices[0].message.content
print(answer_m)
````

````python
# [자가채점]
assert limit == 25
assert isinstance(answer_m, str) and len(answer_m.strip()) > 0
print('✅ 통과!')
````

### 해설

`max_tokens` 는 답이 25토큰에서 끊기게 합니다. 길이를 확실히 줄여 요금·지연을 통제할 때 씁니다.

## 7. 사고모드 — reasoning_effort 로 추론 모델 부르기
**배경**: 추론 모델(`gpt-5-mini`)은 답 전에 속으로 생각합니다. `reasoning_effort` 로 그 깊이를 정합니다.

**요구사항**:
- 사고 수준을 변수 **`effort = 'low'`** 에 담고, `model='gpt-5-mini'`, **`reasoning_effort=effort`** 로 `user='연필 5자루가 2000원이면 12자루는 얼마인가요? 숫자만.'` 를 호출하세요.
- 답변 문자열을 변수 **`answer_reason`** 에 담으세요.

<details><summary>힌트</summary>

```text
접근방법:
- 사고 수준을 변수에 담아 create 의 reasoning_effort 인자로 넘긴다(모델은 gpt-5-mini).

세부구현:
1. effort 에 'low' 를 담는다.
2. create(model='gpt-5-mini', messages, reasoning_effort=effort) 로 호출한다.
3. content 를 answer_reason 에 담아 출력한다.
```

</details>

````python
effort = 'low'
# reasoning_effort 는 gpt-5 계열만 받는다(gpt-4o-mini 에 주면 오류가 난다).
resp = client.chat.completions.create(
    model='gpt-5-mini',
    messages=[{'role': 'user', 'content': '연필 5자루가 2000원이면 12자루는 얼마인가요? 숫자만.'}],
    # low = 생각을 짧게. 한 번에 풀리는 계산이라 굳이 오래 생각할 이유가 없다(빠르고 싸다).
    reasoning_effort=effort)
answer_reason = resp.choices[0].message.content
print(answer_reason)
````

````python
# [자가채점]
assert effort == 'low'
assert isinstance(answer_reason, str) and len(answer_reason.strip()) > 0
print('✅ 통과!')
````

### 해설

간단한 계산이라 `'low'` 로 충분합니다. 어려운 문제일수록 `'high'` 로 올리되 느리고 비싸지는 것을 감안합니다.

## 8. 사고모드 — 어려운 문제엔 높은 effort로
**배경**: 여러 단계를 거쳐야 하는 **어려운 문제**는 `reasoning_effort` 를 **`'high'`** 로 높여 더 깊이 생각하게 합니다(7번은 `'low'`, 8번은 `'high'` — 같은 개념을 **다른 effort 값**으로 비교).

**요구사항**:
- 사고 수준을 변수 **`effort = 'high'`** 에 담고, `model='gpt-5-mini'`, **`reasoning_effort=effort`** 로 아래 논리 문제를 풀게 하세요.
  `user='어떤 수에 3을 곱하고 4를 더했더니 25가 되었습니다. 처음 수는? 숫자만.'`
- 답변 문자열을 변수 **`answer_reason2`** 에 담으세요.

<details><summary>힌트</summary>

```text
접근방법:
- 7번과 같은 구조인데 effort 를 'high' 로 바꾼다.

세부구현:
1. effort 에 'high' 를 담는다.
2. 논리 문제를 user 로 create(model='gpt-5-mini', reasoning_effort=effort) 호출.
3. content 를 answer_reason2 에 담아 출력한다.
```

## 11. 구조화된 출력 — pydantic 스키마 직접 만들기
**배경**: 10번은 스키마를 딕셔너리로 받았습니다(방법 1). 이번엔 교안에서 본 **방법 2(pydantic)** 로, **스키마를 직접 선언**해 결과를 타입 객체로 받습니다(10번과 같은 '구조화 출력' 개념, 다른 방식).

**요구사항**:
- 교안처럼 `BaseModel` 을 상속한 모델 **`Sentiment`** 를 직접 만드세요. 필드는 두 개:
  - `sentiment` : `Literal['긍정', '부정', '중립']`
  - `reason` : `str`
- **`client.chat.completions.parse`** 로 `reviews.loc[1, 'content']`(두 번째 리뷰)를 분석하세요(`model='gpt-4o-mini'`, `response_format=Sentiment`).
- 결과 객체(`resp.choices[0].message.parsed`)를 변수 **`parsed`** 에 담고, `parsed.sentiment` 와 `parsed.reason` 을 출력하세요.

<details><summary>힌트</summary>

```text
접근방법:
- BaseModel 을 상속한 클래스에 필드를 '이름: 타입' 으로 적는다. Literal 로 값 후보를 못박는다.

세부구현:
1. pydantic 에서 BaseModel, typing 에서 Literal 을 가져온다.
2. class Sentiment(BaseModel): 안에 sentiment 와 reason 필드를 타입으로 선언한다.
3. parse(model, messages=[system, user=두 번째 리뷰], response_format=Sentiment) 를 호출한다.
4. 결과의 parsed 를 꺼내 sentiment·reason 을 출력한다.
```

</details>

````python
from pydantic import BaseModel
from typing import Literal

# 10번의 JSON 스키마를 파이썬 클래스로 옮긴 것 — 훨씬 짧고, 오타가 나면 바로 눈에 띈다.
class Sentiment(BaseModel):
    sentiment: Literal['긍정', '부정', '중립']   # 이 셋 말고는 올 수 없다
    reason: str

resp = client.chat.completions.parse(   # create 가 아니라 parse
    model='gpt-4o-mini',
    messages=[
        {'role': 'system', 'content': '리뷰의 감정을 분석해.'},
        {'role': 'user', 'content': reviews.loc[1, 'content']}],
    response_format=Sentiment)
# 10번과 달리 json.loads 가 필요 없다 — .parsed 가 이미 Sentiment 객체다(점으로 꺼내 쓴다).
parsed = resp.choices[0].message.parsed
print(parsed.sentiment, '|', parsed.reason)
````

````python
# [자가채점]
assert parsed.sentiment in {'긍정', '부정', '중립'}
assert isinstance(parsed.reason, str) and len(parsed.reason.strip()) > 0
print('✅ 통과!')
````

### 해설

pydantic 은 `class 이름(BaseModel):` 아래 **필드를 타입으로 선언**합니다. `Literal[...]` 은 방법 1의 `enum` 처럼 값 후보를 못박고, `parse` 는 결과를 **타입 객체**로 줘 `parsed.sentiment` 처럼 점으로 꺼냅니다. 10번(딕셔너리 스키마)과 11번(pydantic)은 같은 '구조화된 출력'을 두 방식으로 익힌 것입니다.

## 12. 파라미터 되짚기 — top_p 로 후보 좁히기
**배경**: 5번의 `temperature` 말고 **후보 자체를 줄이는** 파라미터가 하나 더 있습니다 — `top_p`. 확률이 높은 후보만 남기므로, 낮추면 **매번 비슷한 답**이 나오는 경향이 있습니다.

**요구사항**:
- 질문은 `'가을에 어울리는 디저트 이름을 하나만 지어줘. 이름만 답해.'` 로 고정합니다.
- **`top_p=0.1`** 로 **3번** 호출해 답 문자열들을 리스트 **`narrow`** 에 담으세요.
- **`top_p=1.0`** 로 **3번** 호출해 답 문자열들을 리스트 **`wide`** 에 담으세요.
- 두 리스트에서 **서로 다른 답이 몇 개**인지(`len(set(...))`) 각각 출력해 비교하세요.

**예시**
```
top_p=0.1 → ['단풍사탕', '단풍사탕', '단풍사탕']      서로 다른 답 1개
top_p=1.0 → ['단풍가루 케이크', '가을밤의 단풍케이크', '단풍사탕']   서로 다른 답 3개
```

> ⚠️ 이건 **확률적인 경향**입니다. 3번만 뽑았으니 어쩌다 뒤집힐 수도 있습니다 — **한두 번 돌려 보고 단정하지 않는 것**도 이 문제에서 배울 점입니다.

<details><summary>힌트</summary>

```text
접근방법:
- 같은 질문을 top_p 만 바꿔 3번씩 호출하고, 답을 리스트에 모아 set 으로 중복을 없앤 개수를 센다.

세부구현:
1. 반복문으로 3번 호출해 content 를 리스트에 담는다(top_p=0.1 → narrow).
2. 같은 방식으로 top_p=1.0 → wide 를 만든다.
3. len(set(narrow)) 와 len(set(wide)) 를 출력해 비교한다.
```

</details>

````python
q12 = '가을에 어울리는 디저트 이름을 하나만 지어줘. 이름만 답해.'

# 한 번만 불러서는 다양성 차이를 볼 수 없다 — 같은 설정으로 3번씩 불러 답이 갈리는지 본다.
def ask_top_p(tp):
    out = []
    for _ in range(3):
        resp = client.chat.completions.create(model='gpt-4o-mini',
            messages=[{'role': 'user', 'content': q12}], top_p=tp)
        out.append(resp.choices[0].message.content.strip())
    return out

narrow = ask_top_p(0.1)   # 상위 10% 후보 안에서만 고른다 → 매번 비슷한 답
wide = ask_top_p(1.0)     # 후보를 제한하지 않는다 → 답이 갈린다
# set 으로 감싸면 중복이 사라지므로, 그 개수가 곧 '서로 다른 답이 몇 개였나'가 된다.
print('top_p=0.1 →', narrow, '  서로 다른 답', len(set(narrow)), '개')
print('top_p=1.0 →', wide, '  서로 다른 답', len(set(wide)), '개')
````

````python
# [자가채점]
assert isinstance(narrow, list) and isinstance(wide, list)
assert len(narrow) == 3 and len(wide) == 3, '각 설정으로 3번씩 호출해야 합니다'
assert all(isinstance(x, str) and x.strip() for x in narrow + wide)
# 좁은 쪽이 넓은 쪽보다 훨씬 다양하면 두 설정을 바꿔 넣었을 가능성이 큽니다
assert len(set(narrow)) <= len(set(wide)) + 1, 'top_p 값을 서로 바꿔 넣지 않았는지 확인하세요'
print('✅ 통과!')
````

### 해설

- **접근법**: `temperature` 와 자리만 다를 뿐 쓰는 법은 같습니다 — `create(..., top_p=값)`. 다만 **의미가 다릅니다.** `temperature` 는 확률 분포를 평평하게/뾰족하게 만들고, `top_p` 는 **확률 낮은 후보를 아예 잘라 버립니다**.
- **자가채점이 느슨한 이유**: 3번 표본으로는 다양성 차이가 **뒤집힐 수 있어** `len(set(narrow)) < len(set(wide))` 같은 단정을 채점에 쓸 수 없습니다. 그래서 **호출 방식**(3번씩·올바른 값)만 검사하고, 경향은 눈으로 확인하게 했습니다. **확률적인 현상을 결정적인 기준으로 채점하면 정상 답안이 떨어집니다** — 평가를 설계할 때 늘 겪는 문제입니다.
- **실무**: 보통 `temperature` 를 먼저 조절하고, 이상한 단어가 튀는 것 자체를 막고 싶을 때만 `top_p` 를 낮춥니다. 둘을 동시에 낮추면 답이 지나치게 뻔해집니다.

## 13. 스트리밍 — 답을 조각으로 받기
**배경**: 긴 답은 완성까지 몇 초가 걸립니다. `stream=True` 로 부르면 답이 **조각(chunk)** 으로 오는 대로 화면에 흘려보낼 수 있습니다(교안 2절).

**요구사항**:
- `user='커피의 역사를 세 문장으로 알려줘.'` 를 `gpt-4o-mini`, **`stream=True`** 로 호출하세요.
- 반복문으로 조각을 돌며 **내용이 있는 조각만**(`chunk.choices[0].delta.content` 가 비어 있지 않은 것) 리스트 **`chunks`** 에 모으고, 화면에도 이어서 출력하세요.
- 조각들을 이어 붙인 전체 문자열을 **`stream_text`** 에 담으세요.

<details><summary>힌트</summary>

```text
접근방법:
- stream=True 로 부르면 응답 대신 조각들의 반복자가 온다. for 문으로 돌면 된다.

세부구현:
1. create(model, messages, stream=True) 결과를 for 문으로 돈다.
2. 각 조각에서 delta 의 content 를 꺼내, 값이 있으면 chunks 에 담고 print(..., end='') 로 출력한다.
3. ''.join(chunks) 를 stream_text 에 담는다.
```

</details>

````python
stream = client.chat.completions.create(model='gpt-4o-mini',
    messages=[{'role': 'user', 'content': '커피의 역사를 세 문장으로 알려줘.'}],
    stream=True)   # 완성된 답 대신, 조각을 계속 내보내는 통로가 돌아온다

chunks = []
for chunk in stream:
    # message 가 아니라 delta — '지금 막 늘어난 부분'만 담겨 온다.
    piece = chunk.choices[0].delta.content
    # 맨 앞·맨 뒤 조각에는 글자 없이 정보만 오기도 해서 None 을 걸러야 한다.
    if piece:
        print(piece, end='', flush=True)   # end='' 로 줄바꿈 없이, flush 로 즉시 화면에
        chunks.append(piece)

stream_text = ''.join(chunks)   # 조각을 이어 붙이면 한 번에 받은 답과 같아진다
print('\n\n조각 수:', len(chunks))
````

````python
# [자가채점]
assert isinstance(chunks, list) and len(chunks) > 1, '조각이 하나뿐이면 stream=True 로 부르지 않은 것입니다'
assert isinstance(stream_text, str) and len(stream_text.strip()) > 0
assert stream_text == ''.join(chunks), '조각을 그대로 이어 붙인 것이어야 합니다'
print('✅ 통과!')
````

### 해설

스트리밍은 **총 시간을 줄이지 않습니다** — 첫 글자가 보이기까지의 시간을 줄여 체감을 바꿉니다. 대신 답이 다 오기 전엔 전체 문자열이 없으므로, **집계·파싱처럼 전체가 필요한 작업에는 맞지 않습니다**. 흔한 실수: `delta.content` 가 `None` 인 조각(맨 처음·맨 끝)을 거르지 않아 `TypeError` 가 나는 것.

## 14. 이미지 분석 1) — 로컬 파일을 넣기
**배경**: 모델은 우리 컴퓨터 안을 못 봅니다. 파일을 **base64 문자열**로 바꿔 함께 보내야 합니다(교안 4교시 1절).

**요구사항**:
- `../../day14_OpenAI_API_활용/data/images/food.jpg` 를 읽어 `data:image/jpeg;base64,...` 형태의 문자열로 만들어 변수 **`data_url`** 에 담으세요.
- `content` 를 **리스트**로 만들어 `{'type': 'text', ...}`(질문: **'이 사진에 무엇이 있나요? 한 문장으로.'**)와 `{'type': 'image_url', 'image_url': {'url': 그 문자열}}` 을 함께 넣어 `gpt-4o-mini` 로 호출하세요(`max_tokens=200`).
- 답변 문자열을 **`answer_img`** 에 담으세요.

<details><summary>힌트</summary>

```text
접근방법:
- open(path,'rb') 로 읽어 base64.b64encode(...).decode() 로 문자열을 만든 뒤 앞에 data URL 머리말을 붙인다.

세부구현:
1. base64 를 import 하고 파일을 바이너리로 읽어 인코딩한다.
2. f'data:image/jpeg;base64,{b64}' 문자열을 만든다.
3. content 리스트에 텍스트 항목과 image_url 항목을 넣어 create 를 호출한다.
4. content 를 answer_img 에 담아 출력한다.
```

</details>

````python
import base64
with open('../../day14_OpenAI_API_활용/data/images/food.jpg', 'rb') as f:   # 'rb' — 이미지는 글자가 아니라 바이트다
    b64 = base64.b64encode(f.read()).decode()
# 머리말(data:image/jpeg;base64,)이 있어야 API 가 이 긴 문자열을 그림으로 알아본다.
data_url = f'data:image/jpeg;base64,{b64}'

resp = client.chat.completions.create(model='gpt-4o-mini',
    # 이미지를 넣을 때 달라지는 것은 content 가 문자열에서 조각 리스트가 된다는 점뿐이다.
    messages=[{'role': 'user', 'content': [
        {'type': 'text', 'text': '이 사진에 무엇이 있나요? 한 문장으로.'},
        {'type': 'image_url', 'image_url': {'url': data_url}}]}],
    max_tokens=200)
answer_img = resp.choices[0].message.content
print(answer_img)
````

````python
# [자가채점]
assert isinstance(answer_img, str) and len(answer_img.strip()) > 0
assert data_url.startswith('data:image/jpeg;base64,'), 'data URL 머리말을 붙여야 합니다'
assert len(data_url) > 1000, '파일을 실제로 읽어 인코딩했는지 확인하세요'
print('✅ 통과!')
````

### 해설

이미지를 넣을 때 달라지는 것은 **`content` 가 문자열에서 리스트가 된다**는 점뿐입니다. 흔한 실수: `image_url` 값에 파일 경로를 그대로 넣는 것 — 모델은 우리 파일 시스템에 접근할 수 없으므로 **base64 로 실어 보내야** 합니다.

## 15. 이미지 분석 2) — 공개 URL 로 넣기
**배경**: 이미 웹에 공개된 사진이라면 **주소만** 넘겨도 됩니다(14번과 같은 개념, 다른 전달 방식).

**요구사항**:
- 아래 공개 이미지 주소를 그대로 쓰세요.

```python
IMG_URL = 'https://upload.wikimedia.org/wikipedia/commons/4/45/A_small_cup_of_coffee.JPG'
```

- 14번과 같은 방식으로 호출하되 `image_url` 의 `url` 에 **이 주소를 그대로** 넣으세요(질문: **'이 사진에 무엇이 있나요? 한 문장으로.'**, `max_tokens=200`).
- 답변 문자열을 **`answer_url`** 에 담으세요.

<details><summary>힌트</summary>

```text
접근방법:
- 14번 코드에서 data URL 자리에 웹 주소를 그대로 넣으면 된다(그 외에는 동일).

세부구현:
1. IMG_URL 변수를 만든다.
2. content 리스트의 image_url 에 IMG_URL 을 넣어 create 를 호출한다.
3. content 를 answer_url 에 담아 출력한다.
```

</details>

````python
IMG_URL = 'https://upload.wikimedia.org/wikipedia/commons/4/45/A_small_cup_of_coffee.JPG'

resp = client.chat.completions.create(model='gpt-4o-mini',
    messages=[{'role': 'user', 'content': [
        {'type': 'text', 'text': '이 사진에 무엇이 있나요? 한 문장으로.'},
        # 13번과 같은 자리에 주소만 넣었다 — 파일을 읽어 인코딩할 필요가 없어 훨씬 짧다.
        #  대신 인터넷에 공개된 그림이어야 하고, 그 주소가 사라지면 호출도 실패한다.
        {'type': 'image_url', 'image_url': {'url': IMG_URL}}]}],
    max_tokens=200)
answer_url = resp.choices[0].message.content
print(answer_url)
````

````python
# [자가채점]
assert isinstance(answer_url, str) and len(answer_url.strip()) > 0
assert IMG_URL.startswith('http'), '웹 주소를 그대로 넣어야 합니다'
print('✅ 통과!')
````

### 해설

두 방법의 코드는 거의 같고, **`url` 자리에 무엇을 넣느냐**만 다릅니다. URL 방식은 요청이 가볍지만 **모델 서버가 그 주소에 접속할 수 있어야** 합니다 — 사내망 주소나 로그인이 필요한 주소는 실패합니다. 그래서 업무에서는 대부분 14번(파일) 방식을 씁니다.
